```python
import serial
import time
```

```python

with serial.Serial("/dev/ttyACM3", 115200, timeout=1) as ser:
    for _ in range(5000):
        line = ser.readline()
        time.sleep(0.05)
        if line != b'':
            print(str(line))
```

```python
for _ in range(1000):
    with serial.Serial("/dev/ttyACM3", 115200, timeout=1) as ser:
        if ser.inWaiting() > 0: 
            data = ser.read(ser.inWaiting())
            print(data)
            break
        
        time.sleep(0.01)
```

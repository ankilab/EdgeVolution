```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
# load results.txt
df = pd.read_csv('results.csv', sep=',')
df.columns
```

```python
# plot box plot test acc for each n_fft value
plt.figure()
df.boxplot(column=['test_acc'], by='n_fft', grid=False)
```

```python
plt.figure()
df.boxplot(column=['test_acc'], by='hop_length', grid=False)
```

```python
plt.figure()
df.boxplot(column=['test_acc'], by='n_mel', grid=False)
```

```python
import plotly.express as px
```

```python
# create plotly parallel categories plot
fig = px.parallel_categories(df, dimensions=['n_fft', 'hop_length', 'n_mel'], color="test_acc", color_continuous_scale=px.colors.sequential.Inferno,)
fig.show()
```
